# Snowflake Objects - Database, Schema, and Table Operations
This notebook demonstrates creating and managing core Snowflake objects: databases, schemas, tables, and basic DML operations.

## 1. Session Setup

```
%%sql -r role_set
USE ROLE SYSADMIN;
```

```
%%sql -r wh_set
USE WAREHOUSE COMPUTE_WH;
```

## 2. Database Creation
A **Database** is a logical container for schemas, tables, views, stored procedures, and functions.

```
CREATE OR REPLACE DATABASE DATAENGINEERING;
```

## 3. Schema Creation
A **Schema** is a collection of tables, views, stored procedures, and functions within a database.

```
%%sql -r db_set
USE DATABASE DATAENGINEERING;
```

```
%%sql -r schema_raw
CREATE SCHEMA RAW;
```

```
%%sql -r schema_transformed
CREATE SCHEMA TRANSFORMED;
```

```
%%sql -r schema_enhanced
CREATE SCHEMA ENHNACED;
```

## 4. Table Creation
A **Table** is a data structure (collection of rows and columns) that can hold structured, unstructured, or semi-structured data.

```
%%sql -r tbl_customer
CREATE TABLE RAW.CUSTOMER(
    CUSTOMER_NAME VARCHAR(200),
    CUSTOMER_ID INT PRIMARY KEY,
    CUSTOMER_ADDRESS VARCHAR
);
```

```
%%sql -r tbl_employee
CREATE TABLE RAW.EMPLOYEE(
    EMPLOYEE_NAME VARCHAR(200),
    EMPLOYEE_ID INT PRIMARY KEY,
    EMPLOYEE_ADDRESS VARCHAR,
    DEPT VARCHAR(200)
);
```

```
%%sql -r desc_customer
DESC TABLE RAW.CUSTOMER;
```

```
%%sql -r all_tables
SHOW TABLES;
```

## 5. Inserting and Querying Data

```
%%sql -r insert1
INSERT INTO RAW.CUSTOMER(CUSTOMER_NAME, CUSTOMER_ID, CUSTOMER_ADDRESS)
VALUES ('AJAY', 24, 'OHIO');
```

```
%%sql -r customers_after_insert1
SELECT * FROM RAW.CUSTOMER;
```

```
%%sql -r insert2
INSERT INTO RAW.CUSTOMER(CUSTOMER_NAME, CUSTOMER_ID, CUSTOMER_ADDRESS)
VALUES ('ADAM', 25, 'WISCONSION');
```

```
%%sql -r customers_after_insert2
SELECT * FROM RAW.CUSTOMER;
```

```
%%sql -r insert3
INSERT INTO RAW.CUSTOMER(CUSTOMER_ID) VALUES (245);
```

```
%%sql -r customers_final
SELECT * FROM RAW.CUSTOMER;
```

## 6. Cleanup and New Database
Drop the DATAENGINEERING database and create a new COUNTRIES database.

```
%%sql -r drop_db
DROP DATABASE DATAENGINEERING;
```

```
%%sql -r countries_db
CREATE DATABASE COUNTRIES;
```

```
%%sql -r drop_public
DROP SCHEMA PUBLIC;
```

```
%%sql -r country_schema
CREATE SCHEMA COUNTRY_DATA;
```
